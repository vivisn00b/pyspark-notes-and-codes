```python
# Spark Session

from pyspark.sql import SparkSession

spark = (
    SparkSession
    .builder
    .appName("Reading from CSV Files")
    .master("local[*]")
    .getOrCreate()
)

spark
```

```python
# Read a csv file into dataframe

df = spark.read.format("csv").load("data/input/emp.csv")
df.printSchema()
#df.show()

# Spark sometimes needs to peek into the file to understand its structure.
# It creates job to identify the metadata to optimize read or other functions that we may write later
```

```python
# Reads first row to extract column names

df = spark.read.format("csv").option("header", True).load("data/input/emp.csv")
df.printSchema()
```

```python
# Spark must scan the data to detect column types
# This triggers a job immediately

df = spark.read.format("csv").option("header", True).option("inferSchema", True).load("data/input/emp.csv")
df.printSchema()
df.show()
```

```python
# Reading with Schema

from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DoubleType, DateType

schema = StructType([
    StructField("employee_id", IntegerType(), True),
    StructField("department_id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("age", IntegerType(), True),
    StructField("gender", StringType(), True),
    StructField("salary", DoubleType(), True),
    StructField("hire_date", DateType(), True)
])

_schema = "employee_id int, department_id int, name string, age int, gender string, salary double, hire_date date"
df_schema = spark.read.format("csv").option("header", True).schema(_schema).load("data/input/emp.csv")
# This will not trigger a job
```

```python
df_schema.printSchema()
df_schema.show()
```

```python
# Handle BAD records - PERMISSIVE (Default mode)

_schema = "employee_id int, department_id int, name string, age int, gender string, salary double, hire_date date, bad_record string"
df_permit = spark.read.format("csv").schema(_schema).option("header", True).load("data/input/emp_new.csv")
# To change name of _corrupt_record column
df_permit = spark.read.format("csv").schema(_schema).option("columnNameOfCorruptRecord", "bad_record").option("header", True).load("data/input/emp_new.csv")
```

```python
df_permit.printSchema()
df_permit.show()
```

```python
# Handle BAD records - DROPMALFORMED

_schema = "employee_id int, department_id int, name string, age int, gender string, salary double, hire_date date"
df_mal = spark.read.format("csv").option("mode", "DROPMALFORMED").schema(_schema).load("data/input/emp_new.csv")
# Silently skip (drop) rows that don’t match the expected schema
```

```python
df_mal.printSchema()
df_mal.show()
```

```python
# Handle BAD records - FAILFAST

_schema = "employee_id int, department_id int, name string, age int, gender string, salary double, hire_date date"
df_fail = spark.read.format("csv").option("header", True).option("mode", "FAILFAST").schema(_schema).load("data/input/emp_new.csv")
# Immediately throw an error when it encounters the first malformed row
```

```python
df_fail.printSchema()
df_fail.show()
```

```python
# Multiple options

_options = {
    "header" : "true",
    "inferSchema" : "true",
    "mode" : "PERMISSIVE"
}

df = (spark.read.format("csv").options(**_options).load("data/input/emp.csv"))
```

```python
df.show()
```
