```python
# ==============================
# 🔧 CORE RESOURCE CONFIG
# ==============================

spark.conf.set("spark.executor.instances", "4")        # Number of executors (parallel workers)
spark.conf.set("spark.executor.cores", "4")            # CPU cores per executor
spark.conf.set("spark.executor.memory", "4G")          # Memory per executor
spark.conf.set("spark.driver.memory", "2G")            # Memory for driver process

spark.conf.set("spark.executor.memoryOverhead", "1G")  # Extra memory for JVM overhead (shuffle, serialization)
spark.conf.set("spark.driver.memoryOverhead", "512M")  # Driver overhead memory

# ==============================
# ⚡ DYNAMIC ALLOCATION (AUTO SCALING)
# ==============================

spark.conf.set("spark.dynamicAllocation.enabled", "true")         # Enable dynamic executor scaling
spark.conf.set("spark.dynamicAllocation.minExecutors", "2")       # Minimum executors
spark.conf.set("spark.dynamicAllocation.maxExecutors", "10")      # Maximum executors
spark.conf.set("spark.dynamicAllocation.initialExecutors", "4")   # Initial executors
spark.conf.set("spark.dynamicAllocation.executorIdleTimeout", "60s")  # Remove idle executors after 60s

# ==============================
# 🔁 PARALLELISM & PARTITIONS
# ==============================

spark.conf.set("spark.default.parallelism", "16")         # Default partitions for RDDs (~2–3x total cores)
spark.conf.set("spark.sql.shuffle.partitions", "200")     # Partitions used in shuffle (VERY IMPORTANT)
spark.conf.set("spark.sql.files.maxPartitionBytes", "128MB")  # Max size per partition when reading files
spark.conf.set("spark.sql.files.openCostInBytes", "4MB")      # Cost model for small files

# ==============================
# 🔥 JOIN & SHUFFLE OPTIMIZATION
# ==============================

spark.conf.set("spark.sql.autoBroadcastJoinThreshold", "10MB")  # Max table size for broadcast join
spark.conf.set("spark.sql.broadcastTimeout", "300")             # Timeout for broadcast (seconds)
spark.conf.set("spark.sql.join.preferSortMergeJoin", "true")    # Prefer sort-merge join

# ==============================
# ⚡ ADAPTIVE QUERY EXECUTION (AQE)
# ==============================

spark.conf.set("spark.sql.adaptive.enabled", "true")           # Enable AQE (auto optimization)
spark.conf.set("spark.sql.adaptive.coalescePartitions.enabled", "true")  # Merge small partitions
spark.conf.set("spark.sql.adaptive.skewJoin.enabled", "true")  # Handle skewed joins
spark.conf.set("spark.sql.adaptive.shuffle.targetPostShuffleInputSize", "64MB")  # Target partition size

# ==============================
# 🧮 MEMORY MANAGEMENT
# ==============================

spark.conf.set("spark.memory.fraction", "0.6")         # Memory for execution + storage
spark.conf.set("spark.memory.storageFraction", "0.5")  # Fraction reserved for caching

# ==============================
# 💾 CACHING
# ==============================

spark.conf.set("spark.cleaner.referenceTracking", "true")  # Clean up unused RDDs/DataFrames

# ==============================
# 📦 SERIALIZATION (IMPORTANT FOR PERFORMANCE)
# ==============================

spark.conf.set("spark.serializer", "org.apache.spark.serializer.KryoSerializer")  # Faster serialization
spark.conf.set("spark.kryo.registrationRequired", "false")  # Require class registration (false = flexible)
spark.conf.set("spark.kryoserializer.buffer.max", "256M")   # Max buffer size

# ==============================
# 🌐 NETWORK & TIMEOUTS
# ==============================

spark.conf.set("spark.network.timeout", "120s")        # Network timeout
spark.conf.set("spark.executor.heartbeatInterval", "20s")  # Executor heartbeat interval
spark.conf.set("spark.rpc.askTimeout", "120s")        # RPC timeout

# ==============================
# 📊 SQL / DATAFRAME OPTIMIZATION
# ==============================

spark.conf.set("spark.sql.inMemoryColumnarStorage.compressed", "true")  # Compress cached data
spark.conf.set("spark.sql.inMemoryColumnarStorage.batchSize", "10000")  # Batch size for caching
spark.conf.set("spark.sql.execution.arrow.pyspark.enabled", "true")     # Enable Arrow (fast Pandas conversion)

# ==============================
# 📁 FILE FORMAT (PARQUET / ORC)
# ==============================

spark.conf.set("spark.sql.parquet.compression.codec", "snappy")  # Compression codec
spark.conf.set("spark.sql.parquet.filterPushdown", "true")       # Predicate pushdown

spark.conf.set("spark.sql.orc.filterPushdown", "true")           # ORC pushdown

# ==============================
# 🧵 SCHEDULING
# ==============================

spark.conf.set("spark.scheduler.mode", "FIFO")   # Scheduling mode (FIFO or FAIR)

# ==============================
# 🧪 DEBUGGING & MONITORING
# ==============================

spark.conf.set("spark.eventLog.enabled", "true")     # Enable event logs
spark.conf.set("spark.eventLog.dir", "/tmp/spark-events")  # Log directory
spark.conf.set("spark.ui.enabled", "true")           # Enable Spark UI

# ==============================
# 🚀 STREAMING (OPTIONAL)
# ==============================

spark.conf.set("spark.streaming.backpressure.enabled", "true")  # Auto adjust ingestion rate
spark.conf.set("spark.streaming.kafka.maxRatePerPartition", "1000")  # Limit Kafka rate

# ==============================
# 🧭 DEPLOYMENT (UNCOMMENT BASED ON ENV)
# ==============================

# --- YARN ---
# spark.conf.set("spark.yarn.queue", "default")

# --- KUBERNETES ---
# spark.conf.set("spark.kubernetes.container.image", "your-image")
# spark.conf.set("spark.kubernetes.namespace", "default")

# --- STANDALONE ---
# spark.conf.set("spark.master", "spark://host:7077")
```
